```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix
y=data['Outcome']
x=data.drop('Outcome',axis=1)
x_train, x_test, y_train, y_test=train_test_split(x,y,test_size=0.25,random_state=42)
s=MinMaxScaler()
x_trained=s.fit_transform(x_train)
x_tested=s.fit_transform(x_test)
l=LogisticRegression()
l.fit(x_trained,y_train)
l_pred_train=l.predict(x_trained)
print("Accuracy on trained data : ",accuracy_score(y_train,l_pred_train))
sns.heatmap(confusion_matrix(y_train,l_pred_train),annot=True)
```

```python
l_pred_test=l.predict(x_tested)
print("Accuracy on tested data : ",accuracy_score(y_test,l_pred_test))
sns.heatmap(confusion_matrix(y_test,l_pred_test),annot=True)
```

```python
die_age=data['Age'][data['Outcome']==1]
nondie_age=data['Age'][data['Outcome']==0]
```

```python
from scipy.stats import ttest_ind
stat,p=ttest_ind(die_age,nondie_age)
alpha=0.05
if p<alpha:
    print('Reject H0')
else:
    print("Accept H0")
```

```python
data['BMI']=data['BMI'].replace(0,data['BMI'].median())
die_age_bmi=data['BMI'][data['Outcome']==1]
nondie_age_bmi=data['BMI'][data['Outcome']==0]
stat,p=ttest_ind(die_age_bmi,nondie_age_bmi)
alpha=0.05
if p<alpha:
    print('Reject H0 for BMI Column')
else:
    print("Accept H0 for BMI Column")
```
