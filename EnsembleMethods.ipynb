### Bagging (Bootstrap Aggeregating)

Bagging is an ensemble technique where multiple models train independently on different random samples of dataset and their predictions are combined to produce a final result.

### Why Bagging?

A single decision tree does the follows:

    1) High variance

    2) Easily overfits

    3) Changes a lot if data changes slightly

Bagging reduces variance and makes predictions more stable

### How bagging works?

### Bootstrap sampling

    1) From the original dataset create multiple new datasets

    2) Each dataset is created by random sampling with replacement

    3) Each bootstrap dataset has same size as the original

    ->Some data points repeat,some data points are left out this is called out of bag samples.

### Train multiple models : 

    1) Train one model for bootstrap dataset

    2) Usually that is decision tree

    3) Each tree is trained independently

    4) Make predictions

    5) For regression take the average,for classification take the majority vote.

### Final output : 

    1) Aggregated prediction is more stable and accurate than a single decision tree

### When should we use the bagging?

    1) When dataset is medium to large

    2) Model is overfitting

    3) Decision tree give unstable result


### Boosting

Boosting is an ensemble technique where models are trained sequentially and each new model focus more on the mistakes made by the previous model.


### Why Boosting?

Boosting converts many weak learners into a strong learner.

### How boosting works?

1) Train first weak model 

2) Identify misclassified or high error points 

3) Give more importance to those points(misclassified) 

4) Train next model focusing on them

5) Repeat

6) Combine all models into a final prediction 

### Boosting types

1) Ada Boost

2) Gradient Boost

3) XG Boost

### Ada Boost

Ada boost mainly focus on misclassified points

### Gradient boost

It focus on residual errors

### XG Boost

It is for the optimized solution

```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor,DecisionTreeClassifier
from sklearn.metrics import r2_score,mean_squared_error
from sklearn.ensemble import GradientBoostingRegressor,AdaBoostRegressor
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
```

```python
df=pd.read_csv("housing.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df['total_bedrooms']=df['total_bedrooms'].fillna(df['total_bedrooms'].mean())
```

```python
df.isna().sum()
```

```python
c=['median_house_value','ocean_proximity']
x=df.drop(columns=c,axis=1)
y=df['median_house_value']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
model=DecisionTreeRegressor()
g=GradientBoostingRegressor(n_estimators=100,learning_rate=0.1,max_depth=3,random_state=42)
```

```python
g.fit(x_train,y_train)
y_pred=g.predict(x_test)
print("R2 Score : ",r2_score(y_test,y_pred))
```

```python
g1=AdaBoostRegressor(n_estimators=100,learning_rate=0.1,random_state=42)
g1.fit(x_train,y_train)
y1_pred=g1.predict(x_test)
print("R2 Score : ",r2_score(y_test,y1_pred))
```

```python
g2=XGBRegressor(n_estimators=100,learning_rate=0.1,random_state=42)
g2.fit(x_train,y_train)
y2_pred=g2.predict(x_test)
print("R2 Score : ",r2_score(y_test,y2_pred))
```
